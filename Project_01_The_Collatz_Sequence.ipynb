## The problem

Take a whole number `n` bigger than 0. Then repeat:

- if `n` is even, the next number is `n / 2`
- if `n` is odd, the next number is `3 * n + 1`

Stop when you reach 1.

**Example** with `n = 6`: 6, 3, 10, 5, 16, 8, 4, 2, 1. This is 8 steps.

Nobody knows if every number reaches 1. You will test many numbers with Python.

**Python hint:** `n % 2 == 0` means `n` is even. Use `n // 2` for the division (it gives a whole number).

## Tasks

**Task 1.** Write `collatz_sequence(n)`. It returns the list of numbers from `n` to 1.

Example: `collatz_sequence(6)` gives `[6, 3, 10, 5, 16, 8, 4, 2, 1]`.

**Task 2.** Write `steps(n)` (the number of steps to reach 1) and `highest(n)` (the biggest number in the list).

Example: `steps(6)` gives `8` and `highest(6)` gives `16`.

**Task 3.** Write `longest_start(limit)`. Look at the starting numbers 1, 2, ..., `limit`. Find the one with the most steps. Return it as `(start, steps)`. If two numbers have the same steps, return the smaller number.

Example: `longest_start(10)` gives `(9, 19)`.

Print the answer for `limit = 1000`.

## Your code

Write your code here. Replace `pass` with your own code. You can add more functions.

```python
def collatz_sequence(n):
    # return the list of numbers from n to 1if n <= 0:
    
    if n <= 0:
             return [] 
    
    sequence = [n]
    while n != 1:
        if n % 2 == 0:
            n = n // 2
        else:
            n = 3 * n + 1
        sequence.append(n)
    return sequence


def steps(n):
    seq = collatz_sequence(n)
    if not seq:
        return 0
    return len(seq) - 1


def highest(n):
    seq = collatz_sequence(n)
    if not seq:
        return 0
    return max(seq)


def longest_start(limit):
    # return (start, steps)
     max_steps = 0
     best_start = 1
    
     for i in range(1, limit+1):
        current_steps = steps(i)
        if current_steps > max_steps:
            max_steps = current_steps
            best_start = i
            
     return (best_start, max_steps)


#Optional
def modified_collatz(start):
    n = start
    for _ in range(100):
        if n == 1:
            return True 
        if n % 2 == 0:
            n = n // 2
        else:
            n = 5 * n + 1
    return False 
    
def modified_text():
   not_reached=[]
   for i in range(1,20):
       seq= modified_collatz(i)
       if not seq:
          not_reached.append(i)
   print(not_reached)
modified_text()           
```
